```python
import yfinance as yf
import matplotlib.pyplot as plt
import talib

data= yf.download("AAPL", start="2020-01-01", end="2026-3-30")
print(data.head())

data["Open"].plot(figsize=(10,5))

plt.title("AAPL Closing Price")
plt.show()

data["MA365"]=data["Close"].rolling(window=365).mean()
data[["Close", "MA365"]].plot(figsize=(10, 5))
plt.title("Close Price and one year Moving Average")
plt.show()

close_prices = data["Close"].squeeze()

data["MA365"] = close_prices.rolling(window=365).mean()
data["RSI"] = talib.RSI(close_prices, timeperiod=14)
print(data[["Close", "MA365", "RSI"]].tail())
```
